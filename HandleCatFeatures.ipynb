### Ordinal Encoder

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

# 1. Create a sample DataFrame with explicit values
df = pd.DataFrame({'Satisfaction': ['Low', 'High', 'Medium', 'Low']})

# 2. Define the categorical order explicitly (from lowest to highest)
# This ensures "Medium" gets a value between "Low" and "High"
ordered_cats = [['Low', 'Medium', 'High']]

# 3. Initialize and fit the encoder
oe = OrdinalEncoder(categories=ordered_cats, dtype=int)
df['Satisfaction_Encoded'] = oe.fit_transform(df[['Satisfaction']])

print(df)
```
