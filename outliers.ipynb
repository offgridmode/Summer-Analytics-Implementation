### Outlier Handling using Z-Score

Z = X-u/std


```python
import numpy as np 
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
dataset = [11,10,12,14,15,14,13,15,102,12,14,17,19,107,10,13,12,14,12,108,12,17,16,14,15,10,13,16,105]
outliers = []
```

```python
def detect_outliers(data):
    threshold = 2
    mean = np.mean(data)
    std = np.std(data)

    for i in data:
        z_score = (i-mean)/std
        if np.abs(z_score)>threshold:
            outliers.append(i)
    return outliers
```

```python
detect_outliers(dataset)
```
